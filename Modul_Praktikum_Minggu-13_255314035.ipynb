```python
#buat program pendek yang menunjukkan dapat dikerjakan dengan numpy dan tanpa numpy
#1. perkalian 2 matrix (silakan buat sendiri matrix 2x2)
#2. sistem persamaan linear
                # 2x + 3Y = 8
                # 3x + 4y = 11
#carilah x dan y 
```

```python
# tanpa numpy

A = [[2,5], 
     [3,4]]
B = [[3,2],
     [4,1]]

hasil = [[0,0],
          [0,0]]

for i in range (2):
    for j in range (2):
        for k in range (2):
            hasil [i][j] += A[i][k] * B[k][j]

for row in hasil:
    print(row)
```

```python
# dengan numpy

a = np.array ([[2,5],
               [3,4]])
b = np.array ([[3,2],
               [4,1]])

hasil1 = np.dot (a, b)
hasil2 = np.multiply (a, b)

print(hasil1)
print(hasil2)
```

```python
# tanpa numpy

a1, b1, c1 = 2,3,8
a2, b2, c2 = 3,4,11

y = (c2 * a1 - c1 * a2)/ (b2 * a1 - b1 * a2)
x = (c1 - b1 * y) / a1

print (x)
print (y)
```
