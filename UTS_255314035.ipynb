```python
# Fungsi Membaca data dari file
def load_data():
    tickets = []
    try:
        with open ("tickets.txt" , "r") as file:
           for line in file:
               if ":" in line:
                kota, harga = line.strip().split(":")
                kota = kota.strip()
                harga = int(harga.strip().replace("Rp.", "").replace(" ", ""))
                tickets.append({"kota": kota, "harga": harga})
    except FileNotFoundError:
        pass
    return tickets
# Fungsi menyimpan data dari file
def save_data (tickets):
    with open ("tickets.txt" , "w") as file:
        for ticket in tickets:
            file.write(f"{ticket['kota']} : Rp. {ticket['harga']}\n")

# Fungsi menambahkan kota tujuan dan harga tiket
def tambah_kota(tickets):
    kota = input("Masukkan Nama Kota    : ")
    try:
        harga = int(input("Masukkan Harga Tiket : "))
        tickets.append({"kota": kota, "harga": harga})
        print()
        print("=== DATA BERHASIL DIMASUKKAN ===")
        save_data(tickets)
    except ValueError:
        print()
        print("<Input Tidak Valid!>")

# Fungsi melihat semua list kota dan harga tiket
def lihat_kota(tickets):
    if not tickets :
        print("Tidak Ada Tiket Tersedia!")
        return
    print("=================================")
    print("=== LIST KOTA DAN HARGA TIKET ===")
    print("=================================")
    for i, ticket in enumerate(tickets, start = 1):
        print(f"{i}.{ticket['kota']} - Rp. {ticket['harga']}")
    print("=================================")

#fungsi mengubah harga tiket yang ada di data
def edit_harga(tickets):
    lihat_kota(tickets)
    try:
        nomor = int(input("Nomor Tiket :"))
        if 1 <= nomor <= len(tickets):
            hargabaru = int(input("Masukkan Harga Baru :"))
            tickets[nomor-1]["harga"] = hargabaru
            save_data(tickets)
            print()
            print("===HARGA BERHASIL DI GANTI===")
        else:
            print()
            print("Tiket Tidak Ada!")
    except ValueError:
        print()
        print("Harus Berupa Angka!!")

# Fungsi membeli tiket sesuai data yang ada
def beli_tiket(tickets):
    lihat_kota(tickets)
    try:
        beli = int(input("Beli Tiket Nomor :"))
        jumlah = int(input("Jumlah Tiket Yang dibeli :"))
        if 1 <= beli <= len(tickets):
            kota = tickets[beli-1]["kota"]
            harga = tickets[beli-1]["harga"]
            total_tiket = harga*jumlah

            print()
            print("=======================================")
            print("------------STRUK PEMBELIAN------------")
            print("=======================================")
            print(f"Tujuan                : {kota}")
            print(f"Harga Tiket (1 Orang) : Rp. {harga}")
            print(f"Jumlah Tiket          : {jumlah}")
            print(f"Total Harga           : Rp. {total_tiket}")
            print("=======================================")
            print()
            print("Transaksi Selesai!. Terimakasih")
        else:
            print()
            print("Jumlah Tiket Tidak Valid!")
    except ValueError:
        print()
        print("<Input Tidak Valid!>")

# Fungsi menghapus data tiket
def hapus_tiket(tickets):
     lihat_kota(tickets)
     try:
        angka = int(input("Pilih Angka Tiket Yang Ingin Dihapus :")) -1
        if 0 <= angka < len(tickets):
            tickets.pop(angka)
            save_data(tickets)
            print()
            print ("===TIKET BERHASIL DIHAPUS===")
        else:
            print()
            print("Nomor Tiket Tidak Valid")
     except ValueError:
        print()
        print ("<Input Tidak Valid!>")

# Fungsi Main
def main():
    tickets = load_data()

    while True:
        print()
        print("===================================")
        print("---RAILWAY TICKET VEMDING SYSTEM---")
        print("===================================")
        print(    "[1]. Tampilkan Semua Tiket")
        print(    "[2]. Tambah Tiket Baru")
        print(    "[3]. Edit Harga Tiket")
        print(    "[4]. Beli Tiket")
        print(    "[5]. Hapus Tiket")
        print(    "[6]. Keluar Menu")
        print("===================================")

        pilih = int(input("Pilih No. :"))
        print("-----------------------------------")
        print()

        if pilih == 1:
            lihat_kota(tickets)
        elif pilih == 2:
            tambah_kota(tickets)
        elif pilih == 3:
            edit_harga(tickets)
        elif pilih == 4:
            beli_tiket(tickets)
        elif pilih == 5:
            hapus_tiket(tickets)
        elif pilih == 6:
            print("TERIMA KASIH SUDAH MENGGUNAKAN LAYANAN KAMI!")
            break
#Program Dijalankan        
if __name__ == "__main__":
    main()
```
